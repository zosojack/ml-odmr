# Analisi preliminare fit metodo NLL

```python
import glob
import os
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# =========================================================================
# 1. SCANSIONE DEI CSV NLL E AGGREGAZIONE METRICHE
# =========================================================================
base_nll_dir = os.path.abspath("../benchmarks/10k/likelihood")
csv_files = glob.glob(
    os.path.join(base_nll_dir, "**", "fit_nll_*.csv"), recursive=True
)

print(f"File CSV NLL trovati ed elaborabili: {len(csv_files)}")

K_points = 167          # Numero canali di sweep (file 10000.mat)
delta_sweep_mhz = 200.0 # Finestra fisica [2.75, 2.95] GHz
soglia_c_mhz = 0.4      # Soglia fisica Yao (~5 K)

records = []

for filepath in csv_files:
    match_c = re.search(r"contrast=([\d\.]+)", filepath)
    match_m = re.search(r"mc_trials=(\d+)", filepath)
    match_n = re.search(r"Nph(\d+)", filepath)

    if not (match_c and match_m and match_n):
        continue

    c_val = float(match_c.group(1))
    m_val = int(match_m.group(1))
    n_ph = float(match_n.group(1))

    # Calcolo analitico SNR di singolo canale
    snr_val = c_val * np.sqrt(n_ph / K_points)

    try:
        df = pd.read_csv(filepath, comment="#")
    except Exception:
        continue

    n_total = len(df)
    if n_total == 0:
        continue

    valid = df[df["converged"].astype(bool)]
    n_conv = len(valid)

    if n_conv > 0:
        err_c_mhz = valid["err_c_norm"] * delta_sweep_mhz
        err_s_mhz = valid["err_s_norm"] * delta_sweep_mhz

        rmse_c_mhz = np.sqrt(np.mean(err_c_mhz**2))
        rmse_s_mhz = np.sqrt(np.mean(err_s_mhz**2))
        mae_c_mhz = np.mean(err_c_mhz)
        mae_s_mhz = np.mean(err_s_mhz)

        n_phys_success = np.sum(err_c_mhz < soglia_c_mhz)
        phys_success_rate = (n_phys_success / n_total) * 100.0
        conv_rate = (n_conv / n_total) * 100.0
        mean_time_ms = valid["fit_time_sec"].mean() * 1e3
    else:
        rmse_c_mhz, rmse_s_mhz = np.nan, np.nan
        mae_c_mhz, mae_s_mhz = np.nan, np.nan
        phys_success_rate, conv_rate = 0.0, 0.0
        mean_time_ms = np.nan

    records.append({
        "contrast": c_val,
        "mc_trials": m_val,
        "n_photons": n_ph,
        "snr": snr_val,
        "rmse_c_mhz": rmse_c_mhz,
        "rmse_s_mhz": rmse_s_mhz,
        "mae_c_mhz": mae_c_mhz,
        "mae_s_mhz": mae_s_mhz,
        "conv_rate": conv_rate,
        "phys_success_rate": phys_success_rate,
        "time_per_fit_ms": mean_time_ms,
    })

df_nll = pd.DataFrame(records)
print(f"Riepilogo NLL completato: {len(df_nll)} configurazioni analizzate.")
print(f"Valori di M presenti nei dati: {sorted(df_nll['mc_trials'].unique())}")

# =========================================================================
# 2. GENERAZIONE DEI GRAFICI COMPARATIVI (POISSON NLL)
# =========================================================================
fig, axes = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle(
    "Benchmark Fit Poisson NLL (iminuit) Multi-Start Monte Carlo (ODMR 167 pt)",
    fontsize=14,
    fontweight="bold",
    y=0.98,
)

available_m = sorted(df_nll["mc_trials"].unique())
colors = plt.cm.viridis(np.linspace(0.1, 0.9, max(len(available_m), 1)))
m_color_map = {m: c for m, c in zip(available_m, colors)}

# -------------------------------------------------------------------------
# PANNELLO 1: RMSE sul Centering c vs SNR (log-log)
# -------------------------------------------------------------------------
ax1 = axes[0, 0]
for m_val in available_m:
    sub = df_nll[df_nll["mc_trials"] == m_val].sort_values(by="snr")
    if len(sub) >= 3:
        grouped = sub.groupby(
            pd.qcut(sub["snr"], q=min(12, len(sub)), duplicates="drop"),
            observed=True,
        )[["snr", "rmse_c_mhz"]].mean()
    else:
        grouped = sub[["snr", "rmse_c_mhz"]]

    ax1.plot(
        grouped["snr"],
        grouped["rmse_c_mhz"],
        marker="o",
        lw=1.8,
        color=m_color_map[m_val],
        label=f"M = {m_val}",
    )

ax1.axhline(
    soglia_c_mhz,
    color="crimson",
    linestyle="--",
    alpha=0.7,
    label=f"Soglia target {soglia_c_mhz} MHz",
)
ax1.set_xscale("log")
ax1.set_yscale("log")
ax1.set_xlabel("Rapporto Segnale-Rumore (SNR)")
ax1.set_ylabel("RMSE Centering $c$ (MHz)")
ax1.set_title("RMSE del Centering vs SNR (NLL)")
ax1.grid(True, which="both", linestyle=":", alpha=0.5)
ax1.legend(fontsize=9)

# -------------------------------------------------------------------------
# PANNELLO 2: Tasso di Successo Fisico (|Δc| < 0.4 MHz) vs SNR
# -------------------------------------------------------------------------
ax2 = axes[0, 1]
for m_val in available_m:
    sub = df_nll[df_nll["mc_trials"] == m_val].sort_values(by="snr")
    if len(sub) >= 3:
        grouped = sub.groupby(
            pd.qcut(sub["snr"], q=min(12, len(sub)), duplicates="drop"),
            observed=True,
        )[["snr", "phys_success_rate"]].mean()
    else:
        grouped = sub[["snr", "phys_success_rate"]]

    ax2.plot(
        grouped["snr"],
        grouped["phys_success_rate"],
        marker="s",
        lw=1.8,
        color=m_color_map[m_val],
        label=f"M = {m_val}",
    )

ax2.set_xscale("log")
ax2.set_ylim(-5, 105)
ax2.set_xlabel("Rapporto Segnale-Rumore (SNR)")
ax2.set_ylabel(f"Success Rate Fisico (% con $|\\Delta c| < {soglia_c_mhz}$ MHz)")
ax2.set_title("Tasso di Successo Fisico vs SNR (NLL)")
ax2.grid(True, which="both", linestyle=":", alpha=0.5)
ax2.legend(fontsize=9, loc="lower right")

# -------------------------------------------------------------------------
# PANNELLO 3: Diminishing Returns — RMSE(c) in funzione di M
# -------------------------------------------------------------------------
ax3 = axes[1, 0]
sample_contrast = 0.1 if 0.1 in df_nll["contrast"].values else df_nll["contrast"].median()
df_c_fixed = df_nll[df_nll["contrast"] == sample_contrast]

selected_nphs = sorted(df_c_fixed["n_photons"].unique())
# Evita duplicati se ci sono meno di 3 livelli di N_ph disponibili
nph_targets = list(dict.fromkeys([
    selected_nphs[0],
    selected_nphs[len(selected_nphs) // 2],
    selected_nphs[-1],
])) if len(selected_nphs) > 0 else []

line_styles = ["-.", "--", "-"]

for n_ph, ls in zip(nph_targets, line_styles):
    sub_nph = df_c_fixed[df_c_fixed["n_photons"] == n_ph].sort_values(by="mc_trials")
    if len(sub_nph) > 1:
        ax3.plot(
            sub_nph["mc_trials"],
            sub_nph["rmse_c_mhz"],
            marker="^",
            lw=2.0,
            linestyle=ls,
            label=f"$N_{{ph}} = {n_ph:.1e}$",
        )

ax3.set_yscale("log")
ax3.set_xlabel("Numero di Tentativi Monte Carlo ($M$)")
ax3.set_ylabel("RMSE Centering $c$ (MHz)")
ax3.set_title(f"Impatto di $M$ sull'Errore (Contrasto = {sample_contrast})")
ax3.grid(True, which="both", linestyle=":", alpha=0.5)
ax3.legend(fontsize=9)

# -------------------------------------------------------------------------
# PANNELLO 4: Costo Computazionale — Tempo per Spettro vs M
# -------------------------------------------------------------------------
ax4 = axes[1, 1]
time_by_m = df_nll.groupby("mc_trials")["time_per_fit_ms"].mean()
ax4.bar(
    [str(m) for m in time_by_m.index],
    time_by_m.values,
    color="#1f77b4",
    alpha=0.75,
    edgecolor="black",
    width=0.5,
)

for i, (m_idx, t_val) in enumerate(time_by_m.items()):
    ax4.text(
        i,
        t_val + max(time_by_m.values) * 0.02,
        f"{t_val:.1f} ms",
        ha="center",
        fontweight="bold",
        fontsize=9,
    )

ax4.set_xlabel("Tentativi Monte Carlo ($M$)")
ax4.set_ylabel("Tempo Medio per Spettro (ms)")
ax4.set_title("Costo Computazionale Fit Poisson NLL (iminuit)")
ax4.set_ylim(0, max(time_by_m.values) * 1.15)
ax4.grid(True, axis="y", linestyle=":", alpha=0.5)

plt.tight_layout()
plt.show()
```
