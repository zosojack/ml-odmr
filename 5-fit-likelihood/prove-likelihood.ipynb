Leggo file, predispongo parametri per il noising poissoniano

`````python
import os
import time
from datetime import datetime
from itertools import product
import h5py
import numpy as np
import pandas as pd
from joblib import Parallel, delayed

from ml_odmr.FitMonteCarlo import FitMonteCarlo
from ml_odmr.PoissonNoise import PoissonNoise

# =========================================================================
# 1. PARAMETRI DELLA GRIGLIA DA ESEGUIRE 
# =========================================================================
list_contrast = [0.8, 0.5, 0.2, 0.1, 0.08, 0.05, 0.02, 0.01]
list_Nph = [3e5, 6e5, 12e5, 24e5, 48e5, 60e5]
list_mc_trials = [1, 5, 10, 20, 40]

mat_file_path = "../data/10k/odmr_easyspin_raw_10000.mat"
# Percorso nella root del workspace (allo stesso livello di 'data/')
base_output_dir = os.path.abspath("../benchmarks/10k")
os.makedirs(base_output_dir, exist_ok=True)

random_seed = 42

# Caricamento dati dal file .mat
with h5py.File(mat_file_path, "r") as f:
    spectra_data = np.array(f["spectra_data"]).T
    mw_freqs = np.array(f["mw_freqs"]).squeeze()
    labels_c = np.array(f["labels_c"]).squeeze()
    labels_s = np.array(f["labels_s"]).squeeze()

f_ghz = np.asarray(mw_freqs, dtype=np.float64)
N_spectra_to_run = len(spectra_data)

noise_gen = PoissonNoise(seed=random_seed)
`````

## PROVA SINGOLO FIT

`````python
import numpy as np
from numba import njit
from iminuit import Minuit

# =========================================================================
# 2. DEFINIZIONE DELLA NLL POISSONIANA (NUMBA + CLOSURE)
# =========================================================================
@njit
def _poisson_nll_numba(f_norm, y_counts, A0, A1, x1, w1, A2, x2, w2):
    """
    Calcola la Negative Log-Likelihood poissoniana (a meno di costanti):
        NLL = sum_i [ lambda_i - y_i * ln(lambda_i) ]
    """
    nll = 0.0
    for i in range(len(f_norm)):
        f = f_norm[i]
        l1 = A1 * (w1 * w1 / ((f - x1) * (f - x1) + w1 * w1))
        l2 = A2 * (w2 * w2 / ((f - x2) * (f - x2) + w2 * w2))
        lam = A0 - l1 - l2
        
        # Protezione numerica: lambda deve essere strettamente positivo
        if lam < 1e-9:
            lam = 1e-9
            
        nll += lam - y_counts[i] * np.log(lam)
    return nll


def create_nll(f_norm: np.ndarray, y_counts: np.ndarray):
    """
    Closure che fissa i dati (f_norm, y_counts) e restituisce la funzione
    costo con la firma esplicita dei 7 parametri per iminuit.
    """
    f_arr = np.ascontiguousarray(f_norm, dtype=np.float64)
    y_arr = np.ascontiguousarray(y_counts, dtype=np.float64)
    
    def nll(A0, A1, x1, w1, A2, x2, w2):
        return _poisson_nll_numba(f_arr, y_arr, A0, A1, x1, w1, A2, x2, w2)
        
    return nll


# =========================================================================
# 3. TEST SU UN SINGOLO SPETTRO RUMOROSO
# =========================================================================
idx = 0
contrast_test = 0.08
n_ph_test = 3e5

# Generazione spettro rumoroso (conteggi reali!)
ideal_sig = noise_gen.ideal_signal(spectra_data[idx], contrast=contrast_test)
y_counts = noise_gen.add_noise(ideal_sig, n_tot=n_ph_test, seed_modifier=idx)

# Normalizzazione frequenze in [0, 1]
f_min, f_max = float(f_ghz.min()), float(f_ghz.max())
delta_f = f_max - f_min
f_norm = (f_ghz - f_min) / delta_f

# Stime preliminari in unità di conteggi fotonici
A0_est = float(np.percentile(y_counts, 85))
dip_depth_est = float(max(A0_est - np.min(y_counts), 1.0))

# Istanziazione della NLL e di Minuit
nll_func = create_nll(f_norm, y_counts)

m = Minuit(
    nll_func,
    A0=A0_est,
    A1=dip_depth_est * 0.5,
    x1=0.40,
    w1=0.025,
    A2=dip_depth_est * 0.5,
    x2=0.60,
    w2=0.025,
)

# Fondamentale: 0.5 per NLL (-ln L), così gli errori di HESSE sono a 1 sigma
m.errordef = Minuit.LIKELIHOOD

# Limiti fisici sui parametri
m.limits["A0"] = (0.5 * A0_est, 1.5 * A0_est)
m.limits["A1"] = (0.0, A0_est)
m.limits["A2"] = (0.0, A0_est)
m.limits["x1"] = (0.0, 1.0)
m.limits["x2"] = (0.0, 1.0)
m.limits["w1"] = (0.005, 0.100)
m.limits["w2"] = (0.005, 0.100)

# Esecuzione minimizzazione
m.migrad()

# Estrazione risultati e confronto con ground truth
x1_fit, x2_fit = sorted([m.values["x1"], m.values["x2"]])
c_fit = (x1_fit + x2_fit) / 2.0
s_fit = x2_fit - x1_fit

print(f"Convergenza valida: {m.valid} | Accurate: {m.accurate}")
print(f"c_true = {labels_c[idx]:.5f} | c_fit = {c_fit:.5f} | err_c = {abs(c_fit - labels_c[idx])*200:.3f} MHz")
print(f"s_true = {labels_s[idx]:.5f} | s_fit = {s_fit:.5f} | err_s = {abs(s_fit - labels_s[idx])*200:.3f} MHz")
print(m)
`````

## PROVA FIT AUTOMATIZZATO

`````python
import os
import time
from datetime import datetime
from itertools import product
import numpy as np
import pandas as pd
from numba import njit
from iminuit import Minuit
from joblib import Parallel, delayed
import gc

# =========================================================================
# 1. SUBSET DI CONFIGURAZIONI PER TEST DI VELOCITÀ E ACCURATEZZA
# =========================================================================
list_contrast = [0.8, 0.5, 0.2, 0.1, 0.08, 0.05, 0.02, 0.01]
list_Nph = [3e5, 6e5, 12e5, 24e5, 48e5, 60e5]
list_mc_trials = [1, 5, 10, 20, 40]

# Cartella dedicata per i benchmark NLL (non sovrascrive benchmarks/10k)
base_nll_dir = os.path.abspath("../benchmarks/10k/likelihood")
os.makedirs(base_nll_dir, exist_ok=True)

# Pre-calcoliamo l'asse frequenze normalizzato una sola volta
f_min, f_max = float(f_ghz.min()), float(f_ghz.max())
delta_f = f_max - f_min
f_norm_global = np.ascontiguousarray((f_ghz - f_min) / delta_f, dtype=np.float64)


# =========================================================================
# 2. GENERATORE GUESS POISSONIANO E WORKER NLL
# =========================================================================
@njit
def random_p0_poisson(A0_est, dip_depth_est):
    """Genera parametri iniziali p0 nello spazio [0, 1] per f e conteggi per A."""
    A0 = np.random.uniform(0.95 * A0_est, 1.05 * A0_est)
    A1 = np.random.uniform(0.1 * dip_depth_est, 1.5 * dip_depth_est)
    x1 = np.random.uniform(0.30, 0.55)
    w1 = np.random.uniform(0.01, 0.05)
    A2 = np.random.uniform(0.1 * dip_depth_est, 1.5 * dip_depth_est)
    x2 = np.random.uniform(0.55, 0.80)
    w2 = np.random.uniform(0.01, 0.05)
    return np.array([A0, A1, x1, w1, A2, x2, w2], dtype=np.float64)


def worker_fit_nll(idx: int, contrast_val: float, n_ph_val: float, m_trials: int) -> dict:
    t0 = time.perf_counter()

    # 1. Generazione segnale rumoroso (conteggi reali)
    y_raw = spectra_data[idx]
    ideal_sig = noise_gen.ideal_signal(y_raw, contrast=contrast_val)
    y_counts = noise_gen.add_noise(
        ideal_sig, n_tot=n_ph_val, seed_modifier=idx
    )

    # 2. Stime preliminari in conteggi fotonici
    A0_est = float(np.percentile(y_counts, 85))
    dip_depth_est = float(max(A0_est - np.min(y_counts), 1.0))

    # 3. Istanziazione unica di Minuit per questo spettro
    nll_func = create_nll(f_norm_global, y_counts)
    m = Minuit(
        nll_func,
        A0=A0_est, A1=dip_depth_est * 0.5, x1=0.40, w1=0.025,
        A2=dip_depth_est * 0.5, x2=0.60, w2=0.025,
    )
    m.errordef = Minuit.LIKELIHOOD
    m.limits["A0"] = (0.5 * A0_est, 1.5 * A0_est)
    m.limits["A1"] = (0.0, A0_est)
    m.limits["A2"] = (0.0, A0_est)
    m.limits["x1"] = (0.0, 1.0)
    m.limits["x2"] = (0.0, 1.0)
    m.limits["w1"] = (0.005, 0.100)
    m.limits["w2"] = (0.005, 0.100)

    best_nll = np.inf
    best_x1, best_x2 = np.nan, np.nan
    converged = False

    # 4. Ciclo multi-start Monte Carlo
    for _ in range(m_trials):
        p0 = random_p0_poisson(A0_est, dip_depth_est)
        m.values = p0
        m.migrad(ncall=2500)

        if m.valid and m.fval < best_nll:
            best_nll = float(m.fval)
            best_x1 = float(m.values["x1"])
            best_x2 = float(m.values["x2"])
            converged = True

    duration = time.perf_counter() - t0
    c_true = float(labels_c[idx])
    s_true = float(labels_s[idx])

    if not converged:
        return {
            "spectrum_id": idx,
            "converged": False,
            "c_true_norm": c_true,
            "s_true_norm": s_true,
            "c_fit_norm": np.nan,
            "s_fit_norm": np.nan,
            "err_c_norm": np.nan,
            "err_s_norm": np.nan,
            "c_fit_mhz": np.nan,
            "s_fit_mhz": np.nan,
            "nll_min": np.inf,
            "fit_time_sec": duration,
        }

    # Ordinamento dip e conversione in unità fisiche
    x1_norm, x2_norm = sorted([best_x1, best_x2])
    c_norm = (x1_norm + x2_norm) / 2.0
    s_norm = x2_norm - x1_norm

    x1_phys = (f_min + x1_norm * delta_f) * 1e3
    x2_phys = (f_min + x2_norm * delta_f) * 1e3
    c_mhz = (x1_phys + x2_phys) / 2.0
    s_mhz = x2_phys - x1_phys

    return {
        "spectrum_id": idx,
        "converged": True,
        "c_true_norm": c_true,
        "s_true_norm": s_true,
        "c_fit_norm": c_norm,
        "s_fit_norm": s_norm,
        "err_c_norm": abs(c_norm - c_true),
        "err_s_norm": abs(s_norm - s_true),
        "c_fit_mhz": c_mhz,
        "s_fit_mhz": s_mhz,
        "nll_min": best_nll,
        "fit_time_sec": duration,
    }


# =========================================================================
# 3. ESECUZIONE SUL SUBSET CON SALVATAGGIO INCREMENTALE
# =========================================================================
all_combinations = list(product(list_mc_trials, list_contrast, list_Nph))
total_combos = len(all_combinations)

print(f"Totale combinazioni nel subset NLL: {total_combos}")
print(f"Spettri per combinazione: {N_spectra_to_run}")
print(f"Cartella di output: '{base_nll_dir}'\n")

t_batch_start = time.time()

with Parallel(n_jobs=4, verbose=0) as parallel:
    
    '''
    for combo_idx, (m_val, c_val, n_ph) in enumerate(all_combinations, start=1):

        sub_dir = os.path.join(
            base_nll_dir,
            f"contrast={c_val}",
            f"mc_trials={m_val}",
        )
        os.makedirs(sub_dir, exist_ok=True)

        filename = f"fit_nll_Nph{int(n_ph)}.csv"
        filepath = os.path.join(sub_dir, filename)

        if os.path.exists(filepath):
            print(f"[{combo_idx}/{total_combos}] SALTO (file esistente): contrast={c_val}/mc_trials={m_val}/{filename}")
            continue

        print(f"[{combo_idx}/{total_combos}] Avvio NLL: C={c_val:<4} | M={m_val:<2} | N_ph={n_ph:.1e} ...", end=" ", flush=True)
        t_combo_start = time.time()

        results_dicts = parallel(
            delayed(worker_fit_nll)(i, c_val, n_ph, m_val) for i in range(N_spectra_to_run)
        )

        combo_wall_time = time.time() - t_combo_start

        df_res = pd.DataFrame(results_dicts)
        n_conv = int(df_res["converged"].sum())
        succ_rate = (n_conv / N_spectra_to_run) * 100.0
        mae_c = df_res["err_c_norm"].mean()
        mae_s = df_res["err_s_norm"].mean()
        mean_t = df_res["fit_time_sec"].mean()

        header_metadata = (
            f"# ODMR Poisson NLL (iminuit) Monte Carlo Fit Benchmark\n"
            f"# Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n"
            f"# Source File: {os.path.basename(mat_file_path)}\n"
            f"# N_spectra: {N_spectra_to_run}\n"
            f"# N_photons: {n_ph:.2e}\n"
            f"# Contrast: {c_val}\n"
            f"# MC_Trials_M: {m_val}\n"
            f"# Base_Seed: {random_seed}\n"
            f"# Converged_Spectra: {n_conv}/{N_spectra_to_run} ({succ_rate:.2f}%)\n"
            f"# Mean_MAE_c_norm: {mae_c:.6f}\n"
            f"# Mean_MAE_s_norm: {mae_s:.6f}\n"
            f"# Mean_Fit_Time_Per_Spectrum_s: {mean_t:.4f}\n"
            f"# Total_Wall_Time_s: {combo_wall_time:.2f}\n"
        )

        with open(filepath, mode="w", encoding="utf-8") as f:
            f.write(header_metadata)
            df_res.to_csv(f, index=False, float_format="%.6f")
        '''
    def worker_chunk_nll(
        indices: np.ndarray,
        spectra_chunk: np.ndarray,
        c_true_chunk: np.ndarray,
        s_true_chunk: np.ndarray,
        contrast_val: float,
        n_ph_val: float,
        m_trials: int,
    ) -> list:
        """Processa un intero blocco di spettri in un'unica chiamata worker."""
        chunk_results = []
        # Istanziamo il generatore localmente nel chunk (leggerissimo)
        local_noise = PoissonNoise(seed=random_seed)

        for k in range(len(indices)):
            idx = int(indices[k])
            t0 = time.perf_counter()

            ideal_sig = local_noise.ideal_signal(spectra_chunk[k], contrast=contrast_val)
            y_counts = local_noise.add_noise(ideal_sig, n_tot=n_ph_val, seed_modifier=idx)

            A0_est = float(np.percentile(y_counts, 85))
            dip_depth_est = float(max(A0_est - np.min(y_counts), 1.0))

            nll_func = create_nll(f_norm_global, y_counts)
            m = Minuit(
                nll_func,
                A0=A0_est, A1=dip_depth_est * 0.5, x1=0.40, w1=0.025,
                A2=dip_depth_est * 0.5, x2=0.60, w2=0.025,
            )
            m.errordef = Minuit.LIKELIHOOD
            m.limits["A0"] = (0.5 * A0_est, 1.5 * A0_est)
            m.limits["A1"] = (0.0, A0_est)
            m.limits["A2"] = (0.0, A0_est)
            m.limits["x1"] = (0.0, 1.0)
            m.limits["x2"] = (0.0, 1.0)
            m.limits["w1"] = (0.005, 0.100)
            m.limits["w2"] = (0.005, 0.100)

            best_nll = np.inf
            best_x1, best_x2 = np.nan, np.nan
            converged = False

            for _ in range(m_trials):
                m.values = random_p0_poisson(A0_est, dip_depth_est)
                m.migrad(ncall=2500)
                if m.valid and m.fval < best_nll:
                    best_nll = float(m.fval)
                    best_x1 = float(m.values["x1"])
                    best_x2 = float(m.values["x2"])
                    converged = True

            duration = time.perf_counter() - t0
            c_true = float(c_true_chunk[k])
            s_true = float(s_true_chunk[k])

            if not converged:
                chunk_results.append({
                    "spectrum_id": idx, "converged": False,
                    "c_true_norm": c_true, "s_true_norm": s_true,
                    "c_fit_norm": np.nan, "s_fit_norm": np.nan,
                    "err_c_norm": np.nan, "err_s_norm": np.nan,
                    "c_fit_mhz": np.nan, "s_fit_mhz": np.nan,
                    "nll_min": np.inf, "fit_time_sec": duration,
                })
            else:
                x1_norm, x2_norm = sorted([best_x1, best_x2])
                c_norm = (x1_norm + x2_norm) / 2.0
                s_norm = x2_norm - x1_norm
                x1_phys = (f_min + x1_norm * delta_f) * 1e3
                x2_phys = (f_min + x2_norm * delta_f) * 1e3

                chunk_results.append({
                    "spectrum_id": idx, "converged": True,
                    "c_true_norm": c_true, "s_true_norm": s_true,
                    "c_fit_norm": c_norm, "s_fit_norm": s_norm,
                    "err_c_norm": abs(c_norm - c_true), "err_s_norm": abs(s_norm - s_true),
                    "c_fit_mhz": (x1_phys + x2_phys) / 2.0,
                    "s_fit_mhz": x2_phys - x1_phys,
                    "nll_min": best_nll, "fit_time_sec": duration,
                })

        return chunk_results


    # =========================================================================
    # ESECUZIONE A CHUNK CON 4 WORKER (RAM SOTTO CONTROLLO)
    # =========================================================================
    N_WORKERS = 4
    N_CHUNKS = 16  # 16 blocchi da 625 spettri ciascuno

    all_indices = np.arange(N_spectra_to_run)
    index_chunks = np.array_split(all_indices, N_CHUNKS)

    all_combinations = list(product(list_mc_trials, list_contrast, list_Nph))
    total_combos = len(all_combinations)
    t_batch_start = time.time()

    for combo_idx, (m_val, c_val, n_ph) in enumerate(all_combinations, start=1):
        sub_dir = os.path.join(base_nll_dir, f"contrast={c_val}", f"mc_trials={m_val}")
        os.makedirs(sub_dir, exist_ok=True)
        filepath = os.path.join(sub_dir, f"fit_nll_Nph{int(n_ph)}.csv")

        if os.path.exists(filepath):
            print(f"[{combo_idx}/{total_combos}] SALTO (esistente): C={c_val}, M={m_val}, N_ph={n_ph:.1e}")
            continue

        print(f"[{combo_idx}/{total_combos}] Avvio NLL: C={c_val:<4} | M={m_val:<2} | N_ph={n_ph:.1e} ...", end=" ", flush=True)
        t_combo_start = time.time()

        # Pool locale: si apre e si chiude ad ogni combinazione liberando la RAM C++/LLVM
        chunks_out = Parallel(n_jobs=N_WORKERS, verbose=0)(
            delayed(worker_chunk_nll)(
                ch,
                spectra_data[ch],
                labels_c[ch],
                labels_s[ch],
                c_val,
                n_ph,
                m_val,
            )
            for ch in index_chunks
        )

        # Appiattiamo la lista di liste
        results_dicts = [item for sublist in chunks_out for item in sublist]
        combo_wall_time = time.time() - t_combo_start

        df_res = pd.DataFrame(results_dicts)
        n_conv = int(df_res["converged"].sum())
        succ_rate = (n_conv / N_spectra_to_run) * 100.0
        mae_c = df_res["err_c_norm"].mean()
        mae_s = df_res["err_s_norm"].mean()
        mean_t = df_res["fit_time_sec"].mean()

        header_metadata = (
            f"# ODMR Poisson NLL (iminuit) Monte Carlo Fit Benchmark\n"
            f"# Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n"
            f"# Source File: {os.path.basename(mat_file_path)}\n"
            f"# N_spectra: {N_spectra_to_run}\n"
            f"# N_photons: {n_ph:.2e}\n"
            f"# Contrast: {c_val}\n"
            f"# MC_Trials_M: {m_val}\n"
            f"# Base_Seed: {random_seed}\n"
            f"# Converged_Spectra: {n_conv}/{N_spectra_to_run} ({succ_rate:.2f}%)\n"
            f"# Mean_MAE_c_norm: {mae_c:.6f}\n"
            f"# Mean_MAE_s_norm: {mae_s:.6f}\n"
            f"# Mean_Fit_Time_Per_Spectrum_s: {mean_t:.4f}\n"
            f"# Total_Wall_Time_s: {combo_wall_time:.2f}\n"
        )

        with open(filepath, mode="w", encoding="utf-8") as f:
            f.write(header_metadata)
            df_res.to_csv(f, index=False, float_format="%.6f")

        elapsed_total_min = (time.time() - t_batch_start) / 60.0
        print(f"Fatto in {combo_wall_time:.1f}s | Conv: {succ_rate:.1f}% | MAE_c: {mae_c:.4f} | MAE_s: {mae_s:.4f} | (Tot: {elapsed_total_min:.1f} min)")

        # Pulizia esplicita della RAM prima della combinazione successiva
        del chunks_out, results_dicts, df_res
        gc.collect()

        elapsed_total_min = (time.time() - t_batch_start) / 60.0
        print(f"Fatto in {combo_wall_time:.1f}s | Conv: {succ_rate:.1f}% | MAE_c: {mae_c:.4f} | MAE_s: {mae_s:.4f} | (Tot: {elapsed_total_min:.1f} min)")

print(f"\nTest NLL completato in {(time.time() - t_batch_start)/60:.1f} minuti!")
`````
