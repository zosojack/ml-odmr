```python
import numpy as np

s = np.load('../data/test/X_spectra.npy')
B = np.load('../data/test/Y_B_field.npy')
```

```python
import matplotlib.pyplot as plt
i=15
f = np.arange(2500, 3250)
x_axis = range(2500, 3250)
plt.figure(figsize=(15, 6), dpi=200)
plt.plot(x_axis, s[i])
plt.title(f"B={B[i]}")
```

## Noising

```python
from ml_odmr.noising import *
```

```python
s_out = add_outliers(s[i])
s_pois = add_poisson_noise(s[i], scale=0.3)
s_poly = add_poly_trend(f, s[i])
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(x_axis, s[i])
ax[0, 0].set_title(f"Original Spectrum (B={B[i]})") 
ax[0, 1].plot(x_axis, s_out)
ax[0, 1].set_title("With Outliers")
ax[1, 0].plot(x_axis, s_pois)
ax[1, 0].set_title("With Poisson Noise")           
ax[1, 1].plot(x_axis, s_poly)
ax[1, 1].set_title("With Polynomial Trend")
plt.tight_layout()
plt.show()  
```
